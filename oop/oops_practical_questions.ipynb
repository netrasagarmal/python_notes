## **Q20 — Class decorator wrapping methods**

```python
def deco(cls):
    old = cls.f
    def new(self):
        print("decorated")
        old(self)
    cls.f = new
    return cls

@deco
class A:
    def f(self):
        print("A")

A().f()
```
